### Part 1: Algorithm Implementations & Verification


```python
calls_A = 0
calls_B = 0

def fib_plain(n: int) -> int:
    global calls_A
    calls_A += 1
    if n == 0 or n == 1:
        return n
    else:
        return fib_plain(n - 1) + fib_plain(n - 2)

fibtable = {}

def fib_memo(n: int) -> int:
    global calls_B
    calls_B += 1
    if n in fibtable:
        return fibtable[n]
    if n == 0 or n == 1:
        val = n
    else:
        val = fib_memo(n - 1) + fib_memo(n - 2)
    fibtable[n] = val
    return val

# Verification
print('--- Verifying n = 0 to 20 ---')
for n in range(21):
    calls_A = 0
    calls_B = 0
    fibtable.clear()
    assert fib_plain(n) == fib_memo(n)
print('Both algorithms match for n = 0 to 20.')
```

### Part 2: Experiments and Visualizations

```python
import sys, time
import pandas as pd
import matplotlib.pyplot as plt

# Experiment (a): Algorithm A
results_A = []
for n in [5, 10, 15, 20, 25, 30, 35]:
    calls_A = 0
    t_start = time.perf_counter()
    fib_plain(n)
    t_elapsed = time.perf_counter() - t_start
    results_A.append({'n': n, 'calls': calls_A, 'time (s)': t_elapsed})
    if t_elapsed > 60.0:
        print(f'Exceeded 60s at n={n}')
        break

df_A = pd.DataFrame(results_A)
print('Algorithm A Results:')
display(df_A)
```

```python
# Experiment (b): Algorithm B
sys.setrecursionlimit(3000)
results_B = []
for n in [10, 50, 100, 200, 500, 1000]:
    calls_B = 0
    fibtable.clear()
    t_start = time.perf_counter()
    fib_memo(n)
    t_elapsed = time.perf_counter() - t_start
    results_B.append({'n': n, 'calls': calls_B, 'time (s)': t_elapsed})

df_B = pd.DataFrame(results_B)
print('\nAlgorithm B Results:')
display(df_B)
```

Part 4: Testing and Fixing for n = 1000


```python
# Fix 1: Raise recursion limit
def fix1_recursion(n):
    sys.setrecursionlimit(max(3000, n + 500))
    memo = {}
    def _f(k):
        if k in memo: return memo[k]
        if k <= 1: return k
        memo[k] = _f(k-1) + _f(k-2)
        return memo[k]
    return _f(n)

# Fix 2: Iterative DP
def fix2_iterative(n):
    if n == 0: return 0
    prev2, prev1 = 0, 1
    for _ in range(2, n + 1):
        prev2, prev1 = prev1, prev2 + prev1
    return prev1

val1 = fix1_recursion(1000)
val2 = fix2_iterative(1000)
assert val1 == val2
print(f'Calculated F(1000). Total digits: {len(str(val1))}')
```
